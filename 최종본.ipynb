```python

# ================================================================
# 응급환자 질환별 골든타임 및 실제 도로 이동거리 분석
#
# 분석 대상
# - 환자 1명
# - 지정 병원 21곳
# - 강남구, 마포구, 영등포구, 종로구
#
# 사용 API
# 1. 국립중앙의료원 전국 응급의료기관 정보 조회
# 2. 서울특별시 실시간 도로 소통 정보
# 3. OSRM: OpenStreetMap 기반 자동차 경로 탐색
#
# 분석 기준
# - 구급차 평균속도: 40km/h
# - 심정지: 10분
# - 중증응급: 30분
# - 뇌졸중·심근경색: 60분
# ================================================================


# ================================================================
# 0. 라이브러리 설치
# ================================================================

import sys
import subprocess
import importlib.util

packages = {
    "requests": "requests",
    "pandas": "pandas",
    "numpy": "numpy",
    "matplotlib": "matplotlib",
    "folium": "folium"
}

for module, package in packages.items():
    if importlib.util.find_spec(module) is None:
        subprocess.check_call(
            [sys.executable, "-m", "pip", "install", package]
        )

import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import folium
import xml.etree.ElementTree as ET
import random
import re
import time
import math

from IPython.display import display, HTML


# ================================================================
# 1. API 인증키 및 분석 설정
# ================================================================

HOSPITAL_API_KEY = "bf53fbe5f9a47f0d6f1ed87a9863285f33a05bfb767d8c758267bd57c1d5d1c6"
SEOUL_TRAFFIC_API_KEY = "6a6156446d62736a393863586b4f63"

HOSPITAL_API_URL = (
    "https://apis.data.go.kr/B552657/"
    "ErmctInfoInqireService/getEgytListInfoInqire"
)

SEOUL_TRAFFIC_URL = "http://openapi.seoul.go.kr:8088"

OSRM_URL = "https://router.project-osrm.org"

# 이전에 정상 응답을 확인한 서울시 교통 링크
TRAFFIC_LINK_ID = "1220003800"

# 분석 설정
AMBULANCE_SPEED = 40
DELAY_FACTOR = 1.3

# 지체 계수 민감도 분석
SENSITIVITY_FACTORS = [
    1.0, 1.2, 1.3, 1.5, 1.8, 2.0
]

# 질환별 골든타임
DISEASES = {
    "심정지": {
        "limit": 10,
        "color": "#E53935"
    },
    "중증응급": {
        "limit": 30,
        "color": "#1E88E5"
    },
    "뇌졸중·심근경색": {
        "limit": 60,
        "color": "#FBC02D"
    }
}

TARGET_DISTRICTS = [
    "강남구",
    "마포구",
    "영등포구",
    "종로구"
]

RANDOM_SEED = 42

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)


# ================================================================
# 2. 지정 병원 21곳
# ================================================================

HOSPITAL_NAMES = [
    "서울대학교병원",
    "서울대학교치과병원",
    "서울적십자병원",
    "강북삼성병원",
    "세란병원",
    "신촌연세병원",
    "가톨릭대학교 여의도성모병원",
    "한림대학교한강성심병원",
    "성애의료재단성애병원",
    "씨엠병원",
    "의료법인인봉의료재단영등포병원",
    "한강수병원",
    "한림대학교 강남성심병원",
    "대림성모병원",
    "명지성모병원",
    "차의과학대학교강남차병원",
    "강남베드로병원",
    "연세대학교의과대학강남세브란스병원",
    "더드림병원",
    "재단법인베스티안재단베스티안서울병원",
    "삼성서울병원"
]

# API에 등록된 병원명과 차이가 있을 때 사용하는 별칭
HOSPITAL_ALIASES = {
    "서울대학교병원": ["서울대병원"],
    "서울대학교치과병원": ["서울대치과병원"],
    "서울적십자병원": ["적십자병원"],
    "강북삼성병원": [],
    "세란병원": [],
    "신촌연세병원": [],
    "가톨릭대학교 여의도성모병원": [
        "여의도성모병원",
        "가톨릭대학교여의도성모병원"
    ],
    "한림대학교한강성심병원": ["한강성심병원"],
    "성애의료재단성애병원": ["성애병원"],
    "씨엠병원": ["CM병원"],
    "의료법인인봉의료재단영등포병원": ["영등포병원"],
    "한강수병원": [],
    "한림대학교 강남성심병원": ["강남성심병원"],
    "대림성모병원": [],
    "명지성모병원": [],
    "차의과학대학교강남차병원": ["강남차병원"],
    "강남베드로병원": [],
    "연세대학교의과대학강남세브란스병원": [
        "강남세브란스병원",
        "연세대학교강남세브란스병원"
    ],
    "더드림병원": [],
    "재단법인베스티안재단베스티안서울병원": [
        "베스티안서울병원"
    ],
    "삼성서울병원": []
}


# ================================================================
# 3. 환자 발생 위치
# ================================================================

# 각 구의 모의 환자 발생 좌표
# 실제 행정구역 경계 자료를 이용한 검증 좌표는 아님.
# 행정구역 경계 내부인지 확인하려면 별도 GIS 경계자료가 필요함.

PATIENT_POINTS = {
    "강남구": [
        (37.5172, 127.0473),
        (37.5045, 127.0490),
        (37.4950, 127.0620),
        (37.5250, 127.0350),
        (37.4880, 127.0320)
    ],

    "마포구": [
        (37.5663, 126.9019),
        (37.5550, 126.9100),
        (37.5530, 126.9350),
        (37.5700, 126.9130),
        (37.5500, 126.9200)
    ],

    "영등포구": [
        (37.5264, 126.8963),
        (37.5180, 126.9050),
        (37.5350, 126.8950),
        (37.5130, 126.9240),
        (37.5250, 126.9150)
    ],

    "종로구": [
        (37.5729, 126.9793),
        (37.5800, 126.9700),
        (37.5750, 126.9900),
        (37.5700, 126.9600),
        (37.5850, 126.9750)
    ]
}


# ================================================================
# 4. 공통 함수
# ================================================================

def normalize_name(name):
    return re.sub(
        r"[^가-힣a-zA-Z0-9]",
        "",
        str(name)
    ).lower()


def get_xml_items(root):
    items = []

    for element in root.iter():

        if element.tag.split("}")[-1] == "item":

            row = {}

            for child in element:
                row[child.tag.split("}")[-1]] = child.text

            items.append(row)

    return items


# ================================================================
# 5. 병원 위치 API 수집
# ==============================================================

def fetch_hospital_api():

    if "병원_위치_API_인증키" in HOSPITAL_API_KEY:
        raise ValueError("병원 API 인증키를 입력하세요.")

    collected = []
    page = 1
    rows_per_page = 100

    print("병원 위치 API 연결 중...")

    while page <= 30:

        params = {
            "serviceKey": HOSPITAL_API_KEY,
            "Q0": "서울특별시",
            "pageNo": page,
            "numOfRows": rows_per_page
        }

        response = requests.get(
            HOSPITAL_API_URL,
            params=params,
            timeout=30
        )

        response.raise_for_status()

        root = ET.fromstring(response.content)

        result_code = None
        result_message = None

        for element in root.iter():

            tag = element.tag.split("}")[-1]

            if tag == "resultCode":
                result_code = element.text

            elif tag == "resultMsg":
                result_message = element.text

        if result_code not in (None, "00"):
            raise RuntimeError(
                f"병원 API 오류: {result_code} / {result_message}"
            )

        items = get_xml_items(root)

        if not items:
            break

        collected.extend(items)

        if len(items) < rows_per_page:
            break

        page += 1

    if not collected:
        raise RuntimeError(
            "병원 API에서 위치정보를 가져오지 못했습니다."
        )

    df = pd.DataFrame(collected)

    required = [
        "dutyName",
        "dutyAddr",
        "wgs84Lat",
        "wgs84Lon"
    ]

    for col in required:
        if col not in df.columns:
            raise RuntimeError(
                f"API 응답에 {col} 항목이 없습니다."
            )

    df["wgs84Lat"] = pd.to_numeric(
        df["wgs84Lat"],
        errors="coerce"
    )

    df["wgs84Lon"] = pd.to_numeric(
        df["wgs84Lon"],
        errors="coerce"
    )

    df = df.dropna(
        subset=[
            "dutyName",
            "dutyAddr",
            "wgs84Lat",
            "wgs84Lon"
        ]
    )

    df = df.drop_duplicates(
        subset=["dutyName", "dutyAddr"]
    ).reset_index(drop=True)

    return df


hospital_api_df = fetch_hospital_api()

print(
    "서울시 API에서 수집한 기관 수:",
    len(hospital_api_df)
)


# ================================================================
# 6. 지정 병원과 API 병원명 대조
# ==============================================================

def match_hospitals(api_df):

    matched = []
    unmatched = []

    api_records = api_df.to_dict("records")

    for target in HOSPITAL_NAMES:

        aliases = (
            [target]
            + HOSPITAL_ALIASES.get(target, [])
        )

        aliases = [
            normalize_name(x)
            for x in aliases
        ]

        candidates = []

        for record in api_records:

            api_name = normalize_name(
                record["dutyName"]
            )

            matched_aliases = [
                alias for alias in aliases
                if alias == api_name
            ]

            if matched_aliases:
                candidates.append(record)

        # 정확히 일치하는 항목이 하나일 때만 사용
        if len(candidates) == 1:

            record = candidates[0]

            matched.append({
                "지정병원명": target,
                "API병원명": record["dutyName"],
                "주소": record["dutyAddr"],
                "위도": float(record["wgs84Lat"]),
                "경도": float(record["wgs84Lon"])
            })

        elif len(candidates) > 1:

            unmatched.append({
                "지정병원명": target,
                "상태": "동일 이름 후보가 여러 개"
            })

        else:

            unmatched.append({
                "지정병원명": target,
                "상태": "API에서 정확한 이름 확인 불가"
            })

    return (
        pd.DataFrame(matched),
        pd.DataFrame(unmatched)
    )


hospital_df, unmatched_df = match_hospitals(
    hospital_api_df
)

print("\n병원 위치 대조 결과")
print("지정 병원 수:", len(HOSPITAL_NAMES))
print("좌표 확인 병원 수:", len(hospital_df))
print("확인 실패 병원 수:", len(unmatched_df))

display(hospital_df)

if not unmatched_df.empty:
    print("좌표 확인 실패 병원")
    display(unmatched_df)


# ================================================================
# 7. 환자 위치 무작위 생성
# ==============================================================

selected_district = random.choice(
    TARGET_DISTRICTS
)

patient_lat, patient_lon = random.choice(
    PATIENT_POINTS[selected_district]
)

patient = {
    "환자번호": "P-001",
    "발생지역": selected_district,
    "위도": patient_lat,
    "경도": patient_lon
}

print("\n환자 발생 정보")

display(
    pd.DataFrame([patient])
)


# ================================================================
# 8. OSRM 실제 도로 경로 탐색
# ==============================================================

def get_road_route(
    start_lat,
    start_lon,
    end_lat,
    end_lon
):
    """
    OSRM 자동차 경로 탐색

    반환:
    - 도로 이동 거리(km)
    - OSRM 자체 예상 시간(분)
    - 실제 도로 경로 좌표

    OSRM 시간은 실시간 교통정보가 아닌
    도로망 프로파일 기반 추정치임.
    """

    url = (
        f"{OSRM_URL}/route/v1/driving/"
        f"{start_lon},{start_lat};"
        f"{end_lon},{end_lat}"
    )

    params = {
        "overview": "full",
        "geometries": "geojson",
        "steps": "false",
        "alternatives": "false"
    }

    try:

        response = requests.get(
            url,
            params=params,
            timeout=40
        )

        response.raise_for_status()

        data = response.json()

        if data.get("code") != "Ok":
            return None

        if not data.get("routes"):
            return None

        route = data["routes"][0]

        return {
            "도로거리_km": route["distance"] / 1000,
            "OSRM예상시간_분": route["duration"] / 60,
            "경로좌표": route["geometry"]["coordinates"]
        }

    except Exception as e:

        print("경로 탐색 실패:", e)

        return None


# ================================================================
# 9. 21개 병원의 실제 도로 이동거리 계산
# ==============================================================

if hospital_df.empty:

    raise RuntimeError(
        "병원 좌표가 하나도 확인되지 않았습니다. "
        "병원 API 결과를 확인해 주세요."
    )


route_results = []

print("\n실제 도로 경로 탐색 시작")
print("경로 요청 수:", len(hospital_df))


for index, row in hospital_df.iterrows():

    print(
        f"[{index + 1}/{len(hospital_df)}] "
        f"{row['지정병원명']} 경로 탐색 중..."
    )

    route = get_road_route(
        patient["위도"],
        patient["경도"],
        row["위도"],
        row["경도"]
    )

    if route is None:

        route_results.append({
            "지정병원명": row["지정병원명"],
            "API병원명": row["API병원명"],
            "주소": row["주소"],
            "위도": row["위도"],
            "경도": row["경도"],
            "도로거리(km)": np.nan,
            "OSRM예상시간(분)": np.nan,
            "경로좌표": None,
            "경로상태": "경로 탐색 실패"
        })

    else:

        route_results.append({
            "지정병원명": row["지정병원명"],
            "API병원명": row["API병원명"],
            "주소": row["주소"],
            "위도": row["위도"],
            "경도": row["경도"],
            "도로거리(km)": route["도로거리_km"],
            "OSRM예상시간(분)": route["OSRM예상시간_분"],
            "경로좌표": route["경로좌표"],
            "경로상태": "성공"
        })

    # 공개 데모 서버에 과도한 요청을 보내지 않도록 대기
    time.sleep(0.3)


analysis_df = pd.DataFrame(route_results)

successful_df = analysis_df[
    analysis_df["경로상태"] == "성공"
].copy()

if successful_df.empty:

    raise RuntimeError(
        "실제 도로 경로를 찾지 못했습니다. "
        "OSRM 서버 연결과 병원 좌표를 확인해 주세요."
    )


# ================================================================
# 10. 구급차 평균속도와 지체 계수로 이동시간 계산
# ==============================================================

# 실제 도로 이동거리를 이용
# 예상시간 = 도로거리 / 평균속도 * 60 * 지체계수

analysis_df["예상이동시간(분)"] = (
    analysis_df["도로거리(km)"]
    / AMBULANCE_SPEED
    * 60
    * DELAY_FACTOR
)


# 예상 이동시간 계산 후 성공한 경로만 다시 추출
successful_df = analysis_df[
    analysis_df["경로상태"] == "성공"
].copy()

# 분석에 필요한 열 확인
print("분석 데이터 열 확인:")
print(successful_df.columns.tolist())

print("정상 경로 수:", len(successful_df))


analysis_df["예상이동시간(분)"] = (
    analysis_df["예상이동시간(분)"].round(2)
)

analysis_df["도로거리(km)"] = (
    analysis_df["도로거리(km)"].round(2)
)

analysis_df["OSRM예상시간(분)"] = (
    analysis_df["OSRM예상시간(분)"].round(2)
)

analysis_df = analysis_df.sort_values(
    "예상이동시간(분)",
    na_position="last"
).reset_index(drop=True)


# ================================================================
# 11. 질환별 골든타임 판정
# ==============================================================

for disease, info in DISEASES.items():

    limit = info["limit"]

    analysis_df[f"{disease}_판정"] = np.where(
        analysis_df["예상이동시간(분)"] <= limit,
        "충족",
        "미충족"
    )

    analysis_df[f"{disease}_시간여유(분)"] = (
        limit - analysis_df["예상이동시간(분)"]
    ).round(2)


print("\n병원별 실제 도로 이동거리 및 골든타임 판정")

display(
    analysis_df[
        [
            "지정병원명",
            "주소",
            "도로거리(km)",
            "OSRM예상시간(분)",
            "예상이동시간(분)",
            "경로상태",
            "심정지_판정",
            "중증응급_판정",
            "뇌졸중·심근경색_판정"
        ]
    ]
)


# ================================================================
# 12. 질환별 접근성 통계
# ==============================================================

summary_rows = []

for disease, info in DISEASES.items():

    limit = info["limit"]

    reachable = successful_df[
        successful_df["예상이동시간(분)"] <= limit
    ]

    count = len(reachable)

    total = len(HOSPITAL_NAMES)

    rate = (
        count / total * 100
        if total > 0
        else 0
    )

    if count > 0:

        fastest = reachable.sort_values(
            "예상이동시간(분)"
        ).iloc[0]

        fastest_name = fastest["지정병원명"]
        fastest_time = fastest["예상이동시간(분)"]

    else:

        fastest_name = "해당 없음"
        fastest_time = np.nan

    summary_rows.append({
        "질환": disease,
        "골든타임(분)": limit,
        "도달 가능 병원 수": count,
        "전체 지정 병원": total,
        "경로 확인 병원": len(successful_df),
        "전체 대비 비율(%)": round(rate, 1),
        "가장 빠른 병원": fastest_name,
        "최단 예상시간(분)": fastest_time
    })


summary_df = pd.DataFrame(summary_rows)

print("\n질환별 병원 접근성 종합 분석")

display(summary_df)


# ================================================================
# 질환별 가까운 병원 순위표
# 실제 도로 이동거리 기준
# ==============================================================

for disease, info in DISEASES.items():

    limit = info["limit"]

    # 실제 도로 경로 탐색에 성공한 병원만 사용
    ranking_df = successful_df.copy()

    # 실제 도로 이동거리 기준 오름차순 정렬
    ranking_df = ranking_df.sort_values(
        by="도로거리(km)",
        ascending=True
    ).reset_index(drop=True)

    # 순위 부여
    ranking_df.insert(
        0,
        "순위",
        range(1, len(ranking_df) + 1)
    )

    # 골든타임 충족 여부
    ranking_df["골든타임 판정"] = np.where(
        ranking_df["예상이동시간(분)"] <= limit,
        "충족",
        "미충족"
    )

    # 시간 여유
    ranking_df["시간 여유(분)"] = (
        limit - ranking_df["예상이동시간(분)"]
    ).round(2)

    print("\n" + "=" * 65)
    print(
        f"{disease} 병원 접근성 순위 "
        f"(골든타임 {limit}분)"
    )
    print("=" * 65)

    display(
        ranking_df[
            [
                "순위",
                "지정병원명",
                "주소",
                "도로거리(km)",
                "예상이동시간(분)",
                "골든타임 판정",
                "시간 여유(분)"
            ]
        ]
    )


# ================================================================
# 13. 실제 도로 경로 지도
# ==============================================================

emergency_map = folium.Map(
    location=[
        patient["위도"],
        patient["경도"]
    ],
    zoom_start=12,
    tiles="OpenStreetMap"
)


# 환자 위치
folium.Marker(
    location=[
        patient["위도"],
        patient["경도"]
    ],
    tooltip="응급환자 발생 위치",
    popup=(
        f"환자번호: P-001<br>"
        f"발생지역: {patient['발생지역']}"
    ),
    icon=folium.Icon(
        color="red",
        icon="user",
        prefix="fa"
    )
).add_to(emergency_map)


# 병원별 마커
for _, row in analysis_df.iterrows():

    popup_text = (
        f"<b>{row['지정병원명']}</b><br>"
        f"주소: {row['주소']}<br>"
        f"실제 도로거리: {row['도로거리(km)']} km<br>"
        f"예상 이동시간: {row['예상이동시간(분)']}분<br>"
        f"경로 상태: {row['경로상태']}<br>"
    )

    for disease, info in DISEASES.items():

        if pd.notna(row["예상이동시간(분)"]):

            status = row[f"{disease}_판정"]

        else:

            status = "경로 확인 불가"

        popup_text += (
            f"{disease}: {status} "
            f"({info['limit']}분 기준)<br>"
        )

    folium.Marker(
        location=[
            row["위도"],
            row["경도"]
        ],
        tooltip=row["지정병원명"],
        popup=folium.Popup(
            popup_text,
            max_width=350
        ),
        icon=folium.Icon(
            color="blue",
            icon="plus",
            prefix="fa"
        )
    ).add_to(emergency_map)


# 질환별 실제 도로 경로 레이어
for disease, info in DISEASES.items():

    layer = folium.FeatureGroup(
        name=f"{disease} ({info['limit']}분 이내)",
        show=True
    )

    reachable = analysis_df[
        (analysis_df["예상이동시간(분)"] <= info["limit"])
        & (analysis_df["경로상태"] == "성공")
    ]

    for _, row in reachable.iterrows():

        # OSRM 좌표는 [경도, 위도]
        # Folium은 [위도, 경도]를 사용
        route_latlon = [
            [coordinate[1], coordinate[0]]
            for coordinate in row["경로좌표"]
        ]

        folium.PolyLine(
            locations=route_latlon,
            color=info["color"],
            weight=5,
            opacity=0.8,
            tooltip=(
                f"{disease} | "
                f"{row['지정병원명']} | "
                f"{row['도로거리(km)']}km | "
                f"{row['예상이동시간(분)']}분"
            )
        ).add_to(layer)

    layer.add_to(emergency_map)


folium.LayerControl(
    collapsed=False
).add_to(emergency_map)


# 범례
legend_html = """
<div style="
position: fixed;
bottom: 25px;
left: 25px;
width: 245px;
z-index: 9999;
background: white;
padding: 12px;
border: 2px solid gray;
border-radius: 7px;
font-size: 13px;
">
<b>질환별 실제 도로 경로</b><br><br>
<span style="color:#E53935;">━━━━</span>
심정지 (10분 이내)<br>
<span style="color:#1E88E5;">━━━━</span>
중증응급 (30분 이내)<br>
<span style="color:#FBC02D;">━━━━</span>
뇌졸중·심근경색 (60분 이내)<br>
🔴 환자 발생 위치<br>
🔵 병원 위치
</div>
"""

emergency_map.get_root().html.add_child(
    folium.Element(legend_html)
)

print("\n실제 도로 경로 지도")

display(emergency_map)


# ================================================================
# 14. 질환별 도달 가능 병원 수 그래프
# ==============================================================

plt.figure(figsize=(10, 6))

bars = plt.bar(
    summary_df["질환"],
    summary_df["도달 가능 병원 수"],
    width=0.55
)

plt.title(
    "Disease-specific Hospital Accessibility",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("질환")
plt.ylabel("도달 가능 병원 수")

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{int(height)}개",
        ha="center",
        va="bottom"
    )

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)

plt.tight_layout()
plt.show()


# ================================================================
# 15. 실제 도로 이동거리 및 예상 이동시간 비교
# ==============================================================

plot_df = successful_df.sort_values(
    "예상이동시간(분)"
)

plt.figure(figsize=(12, 8))

plt.barh(
    plot_df["지정병원명"],
    plot_df["예상이동시간(분)"],
    alpha=0.75
)

for disease, info in DISEASES.items():

    plt.axvline(
        x=info["limit"],
        color=info["color"],
        linestyle="--",
        linewidth=2,
        label=f"{disease} ({info['limit']}분)"
    )

plt.xlabel("예상 이동시간 (분)")
plt.ylabel("병원")

plt.title(
    "Road Travel Time vs Golden Time",
    fontsize=14,
    fontweight="bold"
)

plt.legend()
plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()


# ================================================================
# 16. 지체 계수 민감도 분석
# ==============================================================

sensitivity_rows = []

for factor in SENSITIVITY_FACTORS:

    temp = successful_df.copy()

    temp["변경예상시간(분)"] = (
        temp["도로거리(km)"]
        / AMBULANCE_SPEED
        * 60
        * factor
    )

    row = {
        "지체 계수": factor
    }

    for disease, info in DISEASES.items():

        count = (
            temp["변경예상시간(분)"] <= info["limit"]
        ).sum()

        row[f"{disease} 도달 가능 병원"] = int(count)

    sensitivity_rows.append(row)


sensitivity_df = pd.DataFrame(
    sensitivity_rows
)

print("\n지체 계수 민감도 분석")

display(sensitivity_df)


plt.figure(figsize=(11, 6))

for disease, info in DISEASES.items():

    plt.plot(
        sensitivity_df["지체 계수"],
        sensitivity_df[
            f"{disease} 도달 가능 병원"
        ],
        marker="o",
        linewidth=2.5,
        label=disease
    )

plt.title(
    "Sensitivity Analysis by Delay Factor",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("교통 지체 계수")
plt.ylabel("도달 가능 병원 수")

plt.xticks(SENSITIVITY_FACTORS)

plt.yticks(
    range(0, len(HOSPITAL_NAMES) + 1)
)

plt.grid(
    linestyle="--",
    alpha=0.4
)

plt.legend()
plt.tight_layout()
plt.show()


# ================================================================
# 17. 질환별 시간 여유 분석
# ==============================================================

margin_rows = []

for disease, info in DISEASES.items():

    reachable = successful_df[
        successful_df["예상이동시간(분)"] <= info["limit"]
    ]

    if len(reachable) > 0:

        margin = (
            info["limit"]
            - reachable["예상이동시간(분)"]
        )

        margin_rows.append({
            "질환": disease,
            "골든타임(분)": info["limit"],
            "도달 가능 병원 수": len(reachable),
            "평균 시간 여유(분)": round(margin.mean(), 2),
            "최소 시간 여유(분)": round(margin.min(), 2),
            "최대 시간 여유(분)": round(margin.max(), 2)
        })

    else:

        margin_rows.append({
            "질환": disease,
            "골든타임(분)": info["limit"],
            "도달 가능 병원 수": 0,
            "평균 시간 여유(분)": np.nan,
            "최소 시간 여유(분)": np.nan,
            "최대 시간 여유(분)": np.nan
        })


margin_df = pd.DataFrame(margin_rows)

print("\n질환별 골든타임 시간 여유")

display(margin_df)


# ================================================================
# 18. 서울시 실시간 교통 API 정보
# ==============================================================

def fetch_seoul_traffic():

    if "서울교통소통_API_인증키" in SEOUL_TRAFFIC_API_KEY:
        return None

    url = (
        f"{SEOUL_TRAFFIC_URL}/"
        f"{SEOUL_TRAFFIC_API_KEY}/xml/"
        f"TrafficInfo/1/5/{TRAFFIC_LINK_ID}"
    )

    try:

        response = requests.get(
            url,
            timeout=20
        )

        response.raise_for_status()

        root = ET.fromstring(response.content)

        for element in root.iter():

            if element.tag.split("}")[-1] == "row":

                row = {}

                for child in element:
                    row[
                        child.tag.split("}")[-1]
                    ] = child.text

                return row

        return None

    except Exception as e:

        print("교통 API 오류:", e)
        return None


traffic_data = fetch_seoul_traffic()

print("\n서울시 실시간 교통정보")

if traffic_data:

    print("도로 링크:", TRAFFIC_LINK_ID)
    print("관측 속도:", traffic_data.get("prcs_spd"), "km/h")
    print(
        "통행시간:",
        traffic_data.get("prcs_trv_time"),
        "초"
    )

    print(
        "주의: 특정 도로 링크의 관측값이며 "
        "개별 병원 경로의 실시간 교통상황은 아닙니다."
    )

else:

    print("교통정보 조회에 실패했습니다.")


# ================================================================
# 19. 최종 분석 결과 자동 요약
# ==============================================================

print("\n")
print("=" * 70)
print("응급환자 질환별 실제 도로 이동거리 분석 결과")
print("=" * 70)

print("환자번호:", patient["환자번호"])
print("발생지역:", patient["발생지역"])

print(
    "환자 좌표:",
    round(patient["위도"], 5),
    round(patient["경도"], 5)
)

print("지정 병원 수:", len(HOSPITAL_NAMES))
print("좌표 확인 병원 수:", len(hospital_df))
print("도로 경로 확인 병원 수:", len(successful_df))
print("경로 확인 실패 병원 수:", len(analysis_df) - len(successful_df))

print("-" * 70)

for _, row in summary_df.iterrows():

    print(
        f"{row['질환']} "
        f"({row['골든타임(분)']}분): "
        f"{row['도달 가능 병원 수']}개"
    )

print("-" * 70)

if not successful_df.empty:

    fastest = successful_df.sort_values(
        "예상이동시간(분)"
    ).iloc[0]

    print("가장 빠른 병원:", fastest["지정병원명"])
    print("실제 도로거리:", fastest["도로거리(km)"], "km")
    print("예상 이동시간:", fastest["예상이동시간(분)"], "분")

print("=" * 70)


# ================================================================
# 20. 분석 방법 및 한계
# ==============================================================

display(HTML("""
<h3>분석 방법 및 주의사항</h3>

<ol>
<li>
병원은 사용자가 지정한 21곳으로 제한합니다.
국립중앙의료원 API에서 이름과 좌표가 확인된 병원만
실제 경로 분석에 포함합니다.
</li>

<li>
OSRM은 OpenStreetMap 도로망을 기반으로 자동차 경로를
탐색하고 실제 도로망을 따라 계산한 거리를 반환합니다.
</li>

<li>
구급차 예상 이동시간은 도로 이동거리를 40km/h로 이동한다고
가정하고 교통 지체 계수를 적용한 값입니다.
</li>

<li>
OSRM의 기본 예상시간은 실시간 교통정보가 아닙니다.
서울시 교통 API의 특정 도로 링크 정보 역시 모든 병원 경로에
직접 반영된 것은 아닙니다.
</li>

<li>
환자 위치는 각 구에 미리 지정한 모의 좌표 중 하나를
무작위로 선택합니다. 행정구역 경계 자료를 이용한
정밀 검증은 수행하지 않았습니다.
</li>

<li>
골든타임 충족은 도로망 기반 예상시간이 기준 이내라는 뜻이며,
실제 응급실 수용 가능 여부나 전문 진료 가능 여부를
보장하지 않습니다.
</li>

<li>
OSRM 공개 데모 서버는 이용량 제한이 있을 수 있습니다.
대량 요청이나 반복적인 자동 호출은 피해야 합니다.
</li>
</ol>
"""))

print("\n전체 분석 완료")
```
